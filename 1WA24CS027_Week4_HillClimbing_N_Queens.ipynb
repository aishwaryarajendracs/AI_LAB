```python

# Hill Climbing Algorithm for 4-Queens

initial = [3, 1, 2, 0]



def calculate_cost(state):
    cost = 0

    for i in range(4):
        for j in range(i + 1, 4):

            if state[i] == state[j]:
                cost += 1

            elif abs(state[i] - state[j]) == abs(i - j):
                cost += 1

    return cost


# Generate neighbors by swapping two queens
def get_neighbors(state):
    neighbors = []

    for i in range(4):
        for j in range(i + 1, 4):

            new_state = state.copy()

            new_state[i], new_state[j] = (
                new_state[j], new_state[i]
            )

            neighbors.append((new_state, (i, j)))

    return neighbors


# Hill Climbing Algorithm
def hill_climbing():

    current = initial.copy()

    while True:

        current_cost = calculate_cost(current)

        print("\nCurrent State:", current)
        print("Current Cost:", current_cost)

        if current_cost == 0:
            print("Goal State Reached!")
            break

        neighbors = get_neighbors(current)

        print("\nNeighbor Costs:")

        for state, pair in neighbors:
            print("Swap", pair, ":", state,
                  "Cost =", calculate_cost(state))

        # Select neighbor with minimum cost
        # Break ties using smallest column indices
        best_state, best_pair = min(
            neighbors,
            key=lambda x: (calculate_cost(x[0]), x[1])
        )

        best_cost = calculate_cost(best_state)

        # Stop if no better neighbor exists
        if best_cost >= current_cost:
            print("\nNo better neighbor found.")
            print("Algorithm Terminated.")
            break

        print("\nSelected Neighbor:", best_state)
        print("Selected Cost:", best_cost)

        current = best_state


hill_climbing()
```
