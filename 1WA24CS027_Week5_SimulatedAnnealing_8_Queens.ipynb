```python

# Simulated Annealing for 8-Queens

import random
import math

# Calculate number of attacking pairs
def calculate_cost(state):
    cost = 0

    for i in range(8):
        for j in range(i + 1, 8):

            # Check diagonal attacks
            if abs(state[i] - state[j]) == abs(i - j):
                cost += 1

            # Check same row attacks
            elif state[i] == state[j]:
                cost += 1

    return cost


# Generate a random neighbor by swapping two queens
def get_neighbor(state):
    new_state = state.copy()

    i, j = random.sample(range(8), 2)

    new_state[i], new_state[j] = (
        new_state[j], new_state[i]
    )

    return new_state


# Simulated Annealing Algorithm
def simulated_annealing():

    # Generate random initial state
    current = random.sample(range(8), 8)

    temperature = 100
    cooling_rate = 0.99

    best = current.copy()

    iteration = 0

    while temperature > 0.01 and iteration < 100000:

        current_cost = calculate_cost(current)

        # Check whether goal is reached
        if current_cost == 0:
            break

        # Generate random neighbor
        next_state = get_neighbor(current)

        next_cost = calculate_cost(next_state)

        # Calculate change in energy
        deltaE = current_cost - next_cost

        # Accept better state
        if deltaE > 0:
            current = next_state

        # Accept worse state with probability
        else:
            probability = math.exp(deltaE / temperature)

            if random.random() < probability:
                current = next_state

        # Update best state
        if calculate_cost(current) < calculate_cost(best):
            best = current.copy()

        # Decrease temperature
        temperature = temperature * cooling_rate

        iteration += 1

    # Display result
    print("Final State:", best)
    print("Final Cost:", calculate_cost(best))
    print("Iterations:", iteration)

    if calculate_cost(best) == 0:
        print("Solution Found!")

    else:
        print("Solution not found. Try running again.")

    # Display chessboard
    print("\n8 Queens Board:")

    for row in range(8):
        for col in range(8):
            if best[col] == row:
                print("Q", end=" ")
            else:
                print(".", end=" ")
        print()


simulated_annealing()
```
