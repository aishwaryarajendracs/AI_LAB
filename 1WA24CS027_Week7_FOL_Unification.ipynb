```python

# Unification in First Order Logic

# Check whether a variable occurs in a term
def occurs(var, term, subst):
    term = substitute(term, subst)

    if var == term:
        return True

    if isinstance(term, tuple):
        return any(occurs(var, item, subst) for item in term)

    return False


# Apply substitutions to an expression
def substitute(term, subst):
    if isinstance(term, str):
        if term in subst:
            return substitute(subst[term], subst)
        return term

    if isinstance(term, tuple):
        return tuple(substitute(item, subst) for item in term)

    return term


# Unification algorithm
def unify(x, y, subst=None):

    if subst is None:
        subst = {}

    x = substitute(x, subst)
    y = substitute(y, subst)

    # If both expressions are identical
    if x == y:
        return subst

    # If x is a variable
    if isinstance(x, str) and x[0].islower():

        if occurs(x, y, subst):
            return None

        subst[x] = y
        return subst

    # If y is a variable
    if isinstance(y, str) and y[0].islower():

        if occurs(y, x, subst):
            return None

        subst[y] = x
        return subst

    # If both are compound expressions
    if isinstance(x, tuple) and isinstance(y, tuple):

        # Check predicate/function names
        if x[0] != y[0]:
            return None

        # Check number of arguments
        if len(x) != len(y):
            return None

        # Unify corresponding arguments
        for a, b in zip(x[1:], y[1:]):
            subst = unify(a, b, subst)

            if subst is None:
                return None

        return subst

    return None


# Example 1 from the PPT
exp1 = ("Eats", "x", "Apple")
exp2 = ("Eats", "Riya", "y")

result = unify(exp1, exp2)

if result is not None:
    print("Unification Successful!")
    print("Substitution:", result)
    print("Unified Expression:", substitute(exp1, result))
else:
    print("Unification Failed!")
```
