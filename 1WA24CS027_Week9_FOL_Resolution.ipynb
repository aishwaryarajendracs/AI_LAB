```python

# First Order Logic - Resolution Algorithm

# Variables used in the expressions
variables = {'x', 'y', 'k', 'w'}


# Apply substitution to a term
def substitute(term, subst):
    while term in subst:
        term = subst[term]
    return term


# Unification algorithm
def unify(a, b, subst):

    a = substitute(a, subst)
    b = substitute(b, subst)

    if a == b:
        return subst

    if a in variables:
        subst[a] = b
        return subst

    if b in variables:
        subst[b] = a
        return subst

    return None


# Apply substitution to a literal
def apply_substitution(literal, subst):

    neg, predicate, args = literal

    new_args = tuple(
        substitute(arg, subst) for arg in args
    )

    return (neg, predicate, new_args)


# Resolution between two clauses
def resolve(c1, c2):

    for i, l1 in enumerate(c1):
        for j, l2 in enumerate(c2):

            neg1, pred1, args1 = l1
            neg2, pred2, args2 = l2

            # Check complementary literals
            if pred1 == pred2 and neg1 != neg2:

                if len(args1) != len(args2):
                    continue

                subst = {}

                # Unify arguments
                for a, b in zip(args1, args2):
                    subst = unify(a, b, subst)

                    if subst is None:
                        break

                if subst is not None:

                    # Remove resolved literals
                    remaining = (
                        c1[:i] + c1[i+1:] +
                        c2[:j] + c2[j+1:]
                    )

                    # Apply substitutions
                    result = [
                        apply_substitution(lit, subst)
                        for lit in remaining
                    ]

                    return result, subst

    return None, None


# Display clause
def display(clause):

    if not clause:
        return "{}"

    result = []

    for neg, pred, args in clause:

        sign = "NOT " if neg else ""

        result.append(
            sign + pred + "(" + ", ".join(args) + ")"
        )

    return " OR ".join(result)


# Knowledge Base in CNF

C1 = [(True, "Food", ("x",)),
      (False, "Likes", ("John", "x"))]

C4 = [(True, "Eats", ("x", "y")),
      (False, "Killed", ("x",)),
      (False, "Food", ("y",))]

C5 = [(False, "Eats", ("Anil", "Peanuts"))]

C6 = [(False, "Alive", ("Anil",))]

C9 = [(True, "Alive", ("k",)),
      (True, "Killed", ("k",))]

# Negation of the query
C10 = [(True, "Likes", ("John", "Peanuts"))]


# Resolution proof
print("FIRST ORDER LOGIC - RESOLUTION\n")

# Step 1
R1, s1 = resolve(C10, C1)
print("Step 1: Resolve C10 and C1")
print("Substitution:", s1)
print("R1 =", display(R1))

# Step 2
R2, s2 = resolve(R1, C4)
print("\nStep 2: Resolve R1 and C4")
print("Substitution:", s2)
print("R2 =", display(R2))

# Step 3
R3, s3 = resolve(R2, C5)
print("\nStep 3: Resolve R2 and C5")
print("Substitution:", s3)
print("R3 =", display(R3))

# Step 4
R4, s4 = resolve(R3, C9)
print("\nStep 4: Resolve R3 and C9")
print("Substitution:", s4)
print("R4 =", display(R4))

# Step 5
R5, s5 = resolve(R4, C6)
print("\nStep 5: Resolve R4 and C6")
print("Substitution:", s5)
print("R5 =", display(R5))


# Check for empty clause
if R5 == []:
    print("\nEmpty clause obtained!")
    print("Contradiction found.")
    print("Query Likes(John, Peanuts) is PROVED.")
else:
    print("\nQuery could not be proved.")
```
