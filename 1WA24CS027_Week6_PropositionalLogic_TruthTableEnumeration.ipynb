```python

from itertools import product

# Evaluate implication
def implies(p, q):
    return (not p) or q


# Knowledge Base
def KB(A, B, C):
    return (A or C) and (B or not C)


# Query
def query(A, B, C):
    return A or B


# Truth Table Enumeration
def truth_table_entails():

    symbols = ['A', 'B', 'C']

    print("A\tB\tC\tKB\tQuery")

    entailment = True

    for values in product([False, True], repeat=3):

        model = dict(zip(symbols, values))

        A = model['A']
        B = model['B']
        C = model['C']

        kb_value = KB(A, B, C)
        query_value = query(A, B, C)

        print(A, B, C, kb_value, query_value, sep="\t")

        # Check whether KB is true and query is false
        if kb_value and not query_value:
            entailment = False

    if entailment:
        print("\nKB entails the query.")
    else:
        print("\nKB does not entail the query.")


truth_table_entails()
```
